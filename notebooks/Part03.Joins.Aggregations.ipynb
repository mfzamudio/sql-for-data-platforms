## Part 3: Joins & Aggregations

Part 2 of this series (Fundamentals) covered how tables relate through foreign keys. This part
covers the two operations that turn related tables into an answer: **JOIN** (combine rows across
tables) and **aggregation** (collapse many rows into a summary).

## Section 1 — JOIN types

- **INNER JOIN** — only rows where the `ON` condition matches in both tables.
- **LEFT JOIN** — all rows from the left table; `NULL` for right-table columns when there's no match.
- **RIGHT JOIN** — not supported in SQLite; simulate by swapping table order and using `LEFT JOIN`.
- **SELF JOIN** — joins a table to itself via aliases (hierarchies, pairs within one entity).
- **CROSS JOIN** — Cartesian product; every left row paired with every right row.

```python
# INNER JOIN — only customers who have placed at least one order
sql("""
SELECT c.name, o.id AS order_id, o.order_date, o.status
FROM customers c
JOIN orders o ON o.customer_id = c.id
ORDER BY c.name
LIMIT 8
""")
```

```python
# LEFT JOIN — every customer, including the 3 with no orders (customers 23-25)
sql("""
SELECT c.id, c.name, o.id AS order_id
FROM customers c
LEFT JOIN orders o ON o.customer_id = c.id
WHERE o.id IS NULL
""")
```

```python
# 3-table JOIN chain — the most common analytics shape
sql("""
SELECT c.name, o.order_date, p.name AS product, oi.quantity, oi.unit_price
FROM customers c
JOIN orders o       ON o.customer_id = c.id
JOIN order_items oi ON oi.order_id   = o.id
JOIN products p     ON oi.product_id = p.id
ORDER BY o.order_date
LIMIT 8
""")
```

```python
# SELF JOIN — pair employees in the same department (excluding self-pairs)
sql("""
SELECT e1.name AS employee_a, e2.name AS employee_b, e1.department
FROM employees e1
JOIN employees e2 ON e1.department = e2.department AND e1.id < e2.id
ORDER BY e1.department
""")
```

## Section 2 — Aggregations: `GROUP BY`, `HAVING`, aggregate functions

Core functions: `COUNT()`, `SUM()`, `AVG()`, `MIN()`, `MAX()`, `ROUND()`.
`COUNT(DISTINCT col)` counts unique non-NULL values only.

**`WHERE` vs. `HAVING`:** `WHERE` filters rows *before* grouping; `HAVING` filters groups *after*
aggregation. You can use both in one query — `WHERE` to exclude rows first, `HAVING` to filter the
resulting groups.

```python
# Revenue by category — JOIN + GROUP BY
sql("""
SELECT p.category,
       COUNT(DISTINCT o.id)                       AS orders,
       ROUND(SUM(oi.quantity * oi.unit_price), 2) AS revenue
FROM order_items oi
JOIN orders o   ON o.id = oi.order_id
JOIN products p ON p.id = oi.product_id
GROUP BY p.category
ORDER BY revenue DESC
""")
```

```python
# HAVING — only categories with revenue above $1,000
sql("""
SELECT p.category, ROUND(SUM(oi.quantity * oi.unit_price), 2) AS revenue
FROM order_items oi
JOIN products p ON p.id = oi.product_id
GROUP BY p.category
HAVING revenue > 1000
ORDER BY revenue DESC
""")
```

```python
# COUNT(DISTINCT ...) — distinct customers per membership tier
sql("""
SELECT membership_level, COUNT(*) AS customers
FROM customers
GROUP BY membership_level
ORDER BY customers DESC
""")
```

## Best Practices — Joins & Aggregations

- Join on indexed/primary-key columns whenever possible — Module 9 (Performance & Tuning) shows
  exactly why this matters for how many rows the engine has to scan.
- Be explicit about JOIN type. A silent `INNER JOIN` where you meant `LEFT JOIN` quietly drops rows
  — usually the customers/products/whatever with *no* activity, which is often exactly what you
  were trying to measure.
- Every non-aggregated column in `SELECT` must appear in `GROUP BY` — most engines enforce this;
  SQLite is lenient here, which can hide a mistake that a stricter warehouse would reject outright.
