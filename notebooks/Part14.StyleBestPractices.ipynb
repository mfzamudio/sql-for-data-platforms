## 1. Naming conventions

- **Consistent casing:** pick `snake_case` (the overwhelming SQL convention) and stay with it —
  don't mix `customerId` and `customer_id` in the same schema.
- **Prefix by layer, not by type:** `stg_`/`fct_`/`dim_` (Part 2/10's dbt project) tells a reader
  *where in the pipeline* a table sits; Hungarian-style prefixes (`tbl_`, `vw_`) tell them nothing
  they can't already see from the schema.
- **Name booleans as questions:** `is_current` (Part 8), `is_terminal_status` (Part 10's macro) —
  not `status_flag` or `current` (ambiguous: current *what*?).

```python
# GOOD: snake_case, layer-prefixed, boolean named as a question
sql("""
SELECT customer_id, city, is_current
FROM dim_customer_scd2
WHERE is_current = 1
""")

# NOT (illustrative only, not run): mixed casing, type-prefixed name, ambiguous boolean
# SELECT customerId, city, currentFlag FROM tbl_customer_snapshot;
```

## 2. Formatting & readability

- One clause per line for anything beyond a trivial query — `SELECT`/`FROM`/`JOIN`/`WHERE`/
  `GROUP BY` each on their own line, as every notebook in this series has done.
- Prefer a CTE (Part 4) over a nested subquery as soon as a query has more than one logical step.
- Comment the **why**, not the what — `-- Grace MOVED cities` (Part 8) tells a reader something
  the SQL itself can't; `-- select all customers` next to `SELECT * FROM customers` tells them
  nothing.

## 3. CTE vs. subquery — when each wins

| Use a CTE when... | Use a subquery when... |
|---|---|
| The logic has 2+ sequential steps (Part 4, Part 10) | It's a single, simple filter (`WHERE x IN (SELECT ...)`) |
| The same computation is referenced more than once | It's used exactly once, inline |
| You want the query to read top-to-bottom | The nesting is genuinely shallow and clearer inline |

Reminder from Part 9: don't assume a CTE referenced twice is computed once — engines differ.

## 4. Safe DML — the hard rule from Part 1

Every `UPDATE`/`DELETE` in this entire series was preceded by a `SELECT` using the identical
`WHERE` clause. This is not optional style — it's the difference between a scoped change and an
accidental full-table wipe.

```python
# The pattern, one more time: preview, then act, with the SAME WHERE both times
preview = sql("SELECT id, name, membership_level FROM customers WHERE membership_level = 'vip'")
print(f"About to update {len(preview)} rows:")
preview
```

```python
execute("UPDATE customers SET membership_level = 'vip' WHERE membership_level = 'vip'")  # no-op, demonstrates the pattern
print("done -- in a real change, the WHERE above would be identical to the SELECT's WHERE")
```

### Transactions for multi-statement changes

Wrap anything that touches more than one table in an explicit transaction (Part 1) so a failure
partway through doesn't leave the database in a half-changed state — Part 8's SCD Type 2 upsert
(an `UPDATE` + an `INSERT`, both needed to succeed together) is exactly this pattern.

```python
# The transaction-wrapped SCD Type 2 upsert from Part 8, run for real against
# dim_customer_scd2 (created in the Setup section above). Uses conn.execute() directly
# (not the execute() helper, which auto-commits every call) so BEGIN/COMMIT actually
# control one transaction boundary around both statements.
conn.execute("BEGIN TRANSACTION")
conn.execute("""
UPDATE dim_customer_scd2 SET end_date = '2024-06-15', is_current = 0
WHERE customer_id = 7 AND is_current = 1
""")
conn.execute("""
INSERT INTO dim_customer_scd2 (customer_id, name, city, effective_date, is_current)
VALUES (7, 'Grace Kim', 'Montreal', '2024-06-15', 1)
""")
conn.commit()

sql("SELECT customer_key, customer_id, city, effective_date, end_date, is_current FROM dim_customer_scd2 WHERE customer_id = 7")
```

## 5. Common Errors & Fixes

Two traps that come up constantly and are worth seeing go wrong before seeing the fix.

```python
# WRONG: COUNT(o.id) after joining order_items counts each order ONCE PER LINE ITEM
sql("""
SELECT c.name, COUNT(o.id) AS orders_wrong, SUM(oi.quantity * oi.unit_price) AS revenue
FROM customers c
JOIN orders o       ON o.customer_id = c.id
JOIN order_items oi ON oi.order_id   = o.id
WHERE c.id = 1
GROUP BY c.name
""")
```

```python
# RIGHT: COUNT(DISTINCT o.id) counts each order once, no matter how many line items it has
sql("""
SELECT c.name, COUNT(DISTINCT o.id) AS orders_right, SUM(oi.quantity * oi.unit_price) AS revenue
FROM customers c
JOIN orders o       ON o.customer_id = c.id
JOIN order_items oi ON oi.order_id   = o.id
WHERE c.id = 1
GROUP BY c.name
""")
```

```python
# WRONG: both tables have an "id" column -- the engine can't tell which one you mean
try:
    sql("SELECT id FROM customers c JOIN orders o ON o.customer_id = c.id")
except Exception as e:
    print(f"Failed, as expected: {e}")
```

```python
# RIGHT: qualify every column that exists on more than one side of a JOIN
sql("SELECT c.id FROM customers c JOIN orders o ON o.customer_id = c.id LIMIT 3")
```

## 6. A short code-review checklist

Before merging any SQL change:

- [ ] Every `UPDATE`/`DELETE` has a matching `SELECT` preview in the PR description or was tested
      against a non-production copy first.
- [ ] No `SELECT *` in anything that runs repeatedly in production (Part 9 — costs both
      performance and, on consumption-priced platforms, money).
- [ ] Multi-statement changes are wrapped in a transaction.
- [ ] New tables/models have at least `unique`/`not_null` tests on their primary key (Part 10).
- [ ] Any hard-coded date/threshold has a comment explaining why that value, specifically.
- [ ] If touching a dimension table, checked whether SCD Type 1 vs. Type 2 (Part 8) is the right
      call for the changed attribute.
