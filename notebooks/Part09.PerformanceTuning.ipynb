```python
# Baseline: how does SQLite plan this 3-table JOIN + filter, with no indexes?
sql("""
EXPLAIN QUERY PLAN
SELECT c.name, SUM(oi.quantity * oi.unit_price) AS spend
FROM customers c
JOIN orders o       ON o.customer_id = c.id
JOIN order_items oi ON oi.order_id   = o.id
WHERE c.city = 'Toronto'
GROUP BY c.name
""")
```

`SCAN` means the engine reads every row of that table to find matches — expensive as tables
grow. Add indexes on the columns used to `JOIN` and filter, then check the same plan again.

```python
execute("CREATE INDEX idx_customers_city ON customers(city)")
execute("CREATE INDEX idx_orders_customer_id ON orders(customer_id)")
execute("CREATE INDEX idx_order_items_order_id ON order_items(order_id)")

sql("""
EXPLAIN QUERY PLAN
SELECT c.name, SUM(oi.quantity * oi.unit_price) AS spend
FROM customers c
JOIN orders o       ON o.customer_id = c.id
JOIN order_items oi ON oi.order_id   = o.id
WHERE c.city = 'Toronto'
GROUP BY c.name
""")
```

`SCAN` on `customers` became a `SEARCH ... USING INDEX idx_customers_city` — the engine jumps
straight to Toronto rows instead of reading the whole table. The JOINs to `orders`/`order_items`
now also `SEARCH` using their new indexes instead of scanning.

**Common anti-patterns that defeat an index:**
- `SELECT *` when you only need 2 columns — most engines still need to read the full row.
- A function wrapped around the indexed column in `WHERE` (`WHERE UPPER(city) = 'TORONTO'`) —
  the index is built on the raw column, so most engines can't use it here (unless you build a
  functional/expression index specifically on `UPPER(city)`).
- Implicit type conversion (comparing a `TEXT` column to an integer literal) — forces the engine to
  convert every row before comparing, which usually also blocks index use.

## Section 2 — Partitioning & clustering (partition pruning, with DuckDB)

SQLite has no native table partitioning. This is exactly the gap **DuckDB** fills for this
series — it can write and read **Hive-style partitioned Parquet** datasets, and its `EXPLAIN`
shows real pruning, the same mechanism BigQuery/Snowflake/Databricks/Fabric all use under
different names (partitioned tables, micro-partitions + clustering keys, Z-order/liquid
clustering — see Part 12).

```python
import duckdb
import os

os.makedirs('/tmp/spdp_orders_partitioned', exist_ok=True)
ddb = duckdb.connect()

# Build a slightly larger synthetic order table so pruning has something to prune
ddb.execute("""
CREATE TABLE orders_big AS
SELECT
    i AS order_id,
    (DATE '2022-01-01' + INTERVAL (i % 1095) DAY) AS order_date,
    ['Toronto','Vancouver','Montreal','Calgary','Ottawa'][1 + (i % 5)] AS city,
    (10 + (i % 490))::DECIMAL(10,2) AS revenue
FROM range(1, 200000) t(i)
""")

# Write it out partitioned by year — mirrors how a lake table is laid out in production
ddb.execute("""
COPY (SELECT *, EXTRACT(year FROM order_date) AS order_year FROM orders_big)
TO '/tmp/spdp_orders_partitioned' (FORMAT PARQUET, PARTITION_BY (order_year), OVERWRITE_OR_IGNORE)
""")
print("partitioned Parquet dataset written: one folder per order_year")
import subprocess
print(subprocess.run(['ls', '/tmp/spdp_orders_partitioned'], capture_output=True, text=True).stdout)
```

```python
# Query WITHOUT a filter on the partition column — every partition must be read
plan_all = ddb.execute("""
EXPLAIN SELECT COUNT(*) FROM read_parquet('/tmp/spdp_orders_partitioned/*/*.parquet', hive_partitioning=true)
""").fetchall()
for row in plan_all:
    print(row[1])
```

```python
# Query WITH a filter on the partition column — DuckDB prunes to just that partition's files
plan_pruned = ddb.execute("""
EXPLAIN SELECT COUNT(*) FROM read_parquet('/tmp/spdp_orders_partitioned/*/*.parquet', hive_partitioning=true)
WHERE order_year = 2023
""").fetchall()
for row in plan_pruned:
    print(row[1])
```

Look for **`Filters:`** in the pruned plan naming `order_year` — DuckDB determined which
partition folders satisfy the filter *before* touching most of the Parquet files, versus scanning
`read_parquet('*/*.parquet')` with no predicate, which must open every file. Every cloud warehouse
does this same trick at the storage layer: **partition by the column you filter on most (usually
date), cluster/sort by your next-most-common filter.**

## Section 3 — Views vs. materialized views

A plain `CREATE VIEW` (Part 1) re-runs its query every time it's selected from — always fresh,
never cached. A **materialized view** runs the query once and stores the result physically,
trading freshness for speed; it needs an explicit or scheduled refresh. SQLite has no native
materialized views — simulate one as a regular table you refresh on a schedule:

```python
# "Materializing" a view: compute once, store as a table, refresh later on your own schedule
execute("""
CREATE TABLE mv_customer_spend AS
SELECT c.id, c.name, SUM(oi.quantity * oi.unit_price) AS lifetime_spend
FROM customers c
JOIN orders o ON o.customer_id = c.id
JOIN order_items oi ON oi.order_id = o.id
GROUP BY c.id, c.name
""")
sql("SELECT * FROM mv_customer_spend ORDER BY lifetime_spend DESC LIMIT 3")
# To "refresh": DROP TABLE mv_customer_spend and re-run the CREATE TABLE AS above.
# Snowflake/BigQuery/Databricks/Fabric all offer real materialized views with automatic,
# incremental refresh - see Part 12.
```

## Section 4 — JOIN order & CTE materialization

Modern query optimizers reorder JOINs for you based on table statistics — in most engines you
don't need to hand-order tables from smallest to largest. What you *do* still control: whether a
CTE is **inlined** (re-evaluated every place it's referenced, like a view) or **materialized**
(computed once, stored, then reused) — this varies by engine and can matter a lot when a CTE is
expensive and referenced more than once.

```python
# A CTE referenced twice — check whether the plan shows it computed once or twice
sql("""
EXPLAIN QUERY PLAN
WITH category_totals AS (
    SELECT p.category, SUM(oi.quantity * oi.unit_price) AS revenue
    FROM order_items oi JOIN products p ON p.id = oi.product_id
    GROUP BY p.category
)
SELECT a.category, a.revenue
FROM category_totals a
JOIN category_totals b ON a.category = b.category
WHERE a.revenue > 500
""")
```

The plan above shows SQLite chose to **materialize** this CTE: `MATERIALIZE category_totals`
appears once, its own `SCAN`/`GROUP BY` subtree runs once, and both outer references (`SCAN a`,
`SEARCH b`) reuse that single result rather than re-running the subquery. Modern SQLite (3.35+)
decides this per-query based on cost, the same way PostgreSQL has since version 12 (older
PostgreSQL materialized every CTE unconditionally unless a subquery was clearly cheaper);
Snowflake/BigQuery/Databricks make the same kind of cost-based call automatically. **If a CTE is
expensive and used more than once, don't assume it's computed once — check the plan, or
materialize it into a temp table explicitly if you need that guarantee.**

## Section 5 — A short note on cost

Everything in this module reduces *bytes scanned* — and bytes scanned is very often exactly what
cloud warehouses bill for (BigQuery: per byte scanned; Snowflake/Databricks: per-second compute,
but scanning less data means less time running). Partition pruning isn't just a latency win — on
a consumption-priced platform, it's a bill reduction of the same magnitude.

## Best Practices — Performance & Tuning

- Index the columns you `JOIN`/`WHERE` on most; check `EXPLAIN` before and after — don't guess.
- Partition by the column you filter on most often (almost always a date); this is the single
  highest-leverage lever on a large table.
- Don't assume a CTE referenced twice is computed once — verify in the plan.
- A materialized view/table needs a refresh strategy — decide on staleness tolerance up front, not
  after someone notices stale numbers.
