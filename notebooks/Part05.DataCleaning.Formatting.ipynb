## Section 1 — Date parsing & formatting

SQLite stores dates as text and provides `date()`, `strftime()`, `julianday()` to work with them.
**PostgreSQL/MySQL/most warehouses** have a real `DATE`/`TIMESTAMP` type and richer functions
(`DATE_TRUNC`, `EXTRACT`, `TO_CHAR`) — see Part 12 for the cross-platform mapping.

```python
sql("""
SELECT order_date,
       strftime('%Y', order_date)      AS year,
       strftime('%m', order_date)      AS month_number,
       strftime('%Y-%m', order_date)   AS year_month,
       CAST(strftime('%m', order_date) AS INTEGER) AS month_as_int
FROM orders
LIMIT 5
""")
```

```python
# Date arithmetic — days since order, and a "days to ship" style calculation
sql("""
SELECT order_date,
       julianday('now') - julianday(order_date) AS days_since_order
FROM orders
ORDER BY order_date DESC
LIMIT 5
""")
```

## Section 2 — Number formatting & rounding

`ROUND(value, decimals)` is standard everywhere. The subtler issue is **when** to round: round for
*display*, never for intermediate calculations — rounding early and re-aggregating compounds error
across rows.

```python
# Round only at the final, display step — not on every intermediate row
sql("""
SELECT p.category,
       AVG(oi.unit_price)            AS avg_price_unrounded,
       ROUND(AVG(oi.unit_price), 2)  AS avg_price_display
FROM order_items oi
JOIN products p ON p.id = oi.product_id
GROUP BY p.category
""")
```

```python
# CAST for type coercion — e.g. forcing integer division to be exact
sql("""
SELECT 7 / 2                       AS integer_division,
       7.0 / 2                     AS float_division,
       CAST(7 AS REAL) / 2         AS explicit_cast_division
""")
```

## Section 3 — Label & string standardization

Real-world text data arrives inconsistent: mixed case, stray whitespace, inconsistent
abbreviations ("USA" / "U.S." / "United States"). `TRIM`, `UPPER`/`LOWER`, and `REPLACE` are the
core tools; a `CASE WHEN` maps many raw variants onto one canonical label.

```python
sql("""
SELECT DISTINCT city,
       TRIM(city)        AS trimmed,
       UPPER(TRIM(city)) AS standardized
FROM customers
ORDER BY city
""")
```

```python
# CASE WHEN — canonicalizing labels (illustrative: mapping city -> region)
sql("""
SELECT city,
       CASE
           WHEN city IN ('Toronto', 'Ottawa')             THEN 'Ontario'
           WHEN city IN ('Montreal')                      THEN 'Quebec'
           WHEN city IN ('Vancouver')                     THEN 'British Columbia'
           WHEN city IN ('Calgary')                       THEN 'Alberta'
           ELSE 'Other'
       END AS region
FROM customers
GROUP BY city
""")
```

## Section 4 — NULL handling

A `NULL` is not the same as an empty string or zero — it means "unknown/absent." `COALESCE`
substitutes a default; comparing directly to `NULL` with `=` always returns `NULL` (neither true
nor false) — use `IS NULL`/`IS NOT NULL` instead.

```python
sql("""
SELECT name, phone, COALESCE(phone, 'not provided') AS phone_display
FROM customers
WHERE phone IS NULL
LIMIT 5
""")
```

## Best Practices — Data Cleaning

- Round for display only; keep full precision through every intermediate aggregation step.
- Standardize labels with an explicit `CASE WHEN` mapping, not ad-hoc `REPLACE` chains — a mapping
  is auditable; a chain of replacements silently misses variants nobody thought to test.
- Never compare to `NULL` with `=`/`!=` — always `IS NULL`/`IS NOT NULL`.
- Push cleaning logic into a view or a dbt staging model (Part 10) so every downstream query reads
  already-clean data instead of repeating the same `CASE WHEN` in ten different places.
